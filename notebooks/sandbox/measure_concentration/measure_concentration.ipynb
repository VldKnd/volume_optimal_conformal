```python
import torch
import math
import matplotlib.pyplot as plt

def get_ball_volume(dimension: int, radius: float):
    n = dimension
    log_volume = (n / 2) * math.log(torch.pi * radius**2) - math.lgamma((n / 2) + 1)
    return math.exp(log_volume)
```

```python
rings_masses = []
rings_volumes = []

sphere_masses = []
sphere_volumes = []

epsilon = 0.1
batch_size = 10024

for n in [2**i for i in range(1, 17)]:
    l_radius = n**(1/2) - epsilon
    r_radius = n**(1/2) + epsilon
    rings_volumes.append(
        get_ball_volume(
            dimension=n,
            radius=r_radius
        ) - \
        get_ball_volume(
            dimension=n,
            radius=l_radius
        )
    )

    sphere_radius = ((r_radius/l_radius)**n - 1)**(1/n) * l_radius
    sphere_volumes.append(
        get_ball_volume(
            dimension=n,
            radius=sphere_radius
        )
    )

    random_vectors = torch.randn(batch_size, n)
    random_vectors_norms = random_vectors.norm(dim=-1, keepdim=True)

    inclusion_in_donut = ((random_vectors_norms > l_radius) * (random_vectors_norms < r_radius)).to(float)
    rings_masses.append(inclusion_in_donut.mean())

    inclusion_in_sphere = (random_vectors_norms < sphere_radius).to(float)
    sphere_masses.append(inclusion_in_sphere.mean())
    
    print(
        f"dimension: {n}\n"
        f"ring's mass: {rings_masses[-1]}\n"
        f"ring's volume: {rings_volumes[-1]}\n\n"
        f"sphere's mass: {sphere_masses[-1]}\n"
        f"sphere's volume: {sphere_volumes[-1]}\n\n"
        f"radiuses mean: {random_vectors_norms.mean()}\n"
        f"radiuses std: {random_vectors_norms.std()}\n\n"
    )
```
